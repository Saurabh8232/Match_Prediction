```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.read_csv("C:\\Users\\mukes\\OneDrive\\Documents\\ML\\ipl data set.csv")
x=df[['batting_team', 'bowling_team', 'venue',
       'pitch_condition', 'current_score', 'overs_completed', 'wickets_out',
       'runs_last_5_overs', 'overs_remaining', 'run_rate']]
y=df["final_score"]


encoder1=OrdinalEncoder()
x['pitch_condition']=encoder1.fit_transform(x[['pitch_condition']])
encoder2=OneHotEncoder(sparse_output=False)
converted_data=encoder2.fit_transform(x[['batting_team','bowling_team','venue']])

column_names=encoder2.get_feature_names_out(['batting_team','bowling_team','venue'])
converted_data_df=pd.DataFrame(converted_data,columns=column_names)

x=x.drop(['batting_team','bowling_team','venue'],axis=1)

x=pd.concat([x.reset_index(drop=True),converted_data_df.reset_index(drop=True)],axis=1)

```

```python
model1=LinearRegression()
poly1=PolynomialFeatures(degree=2)
model2=LinearRegression()

i=1
kf=KFold(n_splits=5,shuffle=True,random_state=42)
for train,test in kf.split(x):
    x_train,x_test=x.iloc[train],x.iloc[test]
    y_train,y_test=y.iloc[train],y.iloc[test]

    model1.fit(x_train,y_train)
    print("LinearRegression model",i,"score: ",model1.score(x_test,y_test))

    x_train_poly_data1=poly1.fit_transform(x_train)    
    x_test_poly_data1=poly1.transform(x_test)
    model2.fit(x_train_poly_data1,y_train)
    print("PolynomialRegression model",i,"degree 2 score: ",model2.score(x_test_poly_data1,y_test))

    
    


    i=i+1


```

```python
score = pd.DataFrame([{
'batting_team': 'RCB',
'bowling_team': 'PBKS',
'venue': 'Rajiv Gandhi Stadium',
'pitch_condition': 'Flat',
'current_score': 125,
'overs_completed': 14,
'wickets_out': 3,
'runs_last_5_overs': 45,
'overs_remaining': 6,
'run_rate': 8.928
}])

score['pitch_condition'] = encoder1.transform(score[['pitch_condition']])

score_converted_data = encoder2.transform(score[['batting_team','bowling_team','venue']])

score_column_names = encoder2.get_feature_names_out(['batting_team','bowling_team','venue'])

score_converted_data_df = pd.DataFrame(score_converted_data, columns=score_column_names)

score = score.drop(['batting_team','bowling_team','venue'], axis=1)

score = pd.concat([score.reset_index(drop=True),score_converted_data_df.reset_index(drop=True)], axis=1)


score_poly1 = poly1.transform(score)

score_predict1 = model2.predict(score_poly1)


print(round(score_predict1[0]))
```
